```python
import pandas as pd
import pickle
import os
import time


OUTPUT_DIR = os.getenv("OUTPUT_DIR", './output')
MODEL_PATH = f"{OUTPUT_DIR}/model.pkl"
TEST_PATH = f"{OUTPUT_DIR}/mission15_test.csv"
RESULT_PATH = f"{OUTPUT_DIR}/result.csv"

print("⏳ model.pkl 생성 대기 중...")

# 1️⃣ model.pkl 생길 때까지 대기
while not os.path.exists(MODEL_PATH):
    time.sleep(1)

print("✅ model.pkl 발견!")

# 2️⃣ 모델 + 스케일러 로드
with open(MODEL_PATH, "rb") as f:
    saved = pickle.load(f)

model = saved["model"]
scaler = saved["scaler"]

# 3️⃣ 테스트 데이터 로드
df_test = pd.read_csv(TEST_PATH)

df_test['Extracurricular Activities'] = df_test['Extracurricular Activities'].map({
    'Yes': 1,
    'No': 0
})

X_test = df_test[['Hours Studied', 'Previous Scores',
                   'Extracurricular Activities',
                   'Sleep Hours',
                   'Sample Question Papers Practiced']]

# ⭐ 반드시 같은 스케일러 사용
X_test_scaled = scaler.transform(X_test)

# 4️⃣ 예측
pred = model.predict(X_test_scaled)

# 5️⃣ 결과 저장
df_test['Prediction'] = pred
df_test.to_csv(RESULT_PATH, index=False)

print("🎉 result.csv 저장 완료!")
```
